```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("tablas_persistentes")
    .enableHiveSupport()
    .getOrCreate()
)

resumen = spark.read.parquet("/opt/datos/resumen_ventas")

spark.sql("CREATE DATABASE IF NOT EXISTS tienda")
resumen.write.mode("overwrite").saveAsTable("tienda.resumen_ventas")
spark.sql("SHOW TABLES IN tienda").show()
```
